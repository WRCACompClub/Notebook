### 1.4 If statements


An "if" statement runs code if something is true.

Notice how below, it says `if 2 + 2 == 4`. The "==" is different from "="because if you say `age = 5`, it means age **IS** equal to 5. Meanwhile, if you say `age == 5`, this would be asking python if "age is equal to 5?", so it would be a true or false statement.
<br>
<br>
Besides if, there is also "*`elif`*", which just mean "else if". That means if the above if statement isn't true, the code will check this "*`elif`*" statement. You can link as many "*`elif`*" statements to an if statement as you want.
<br>

Besides if and elif, there is also "*`else`*", which only runs if all of the above if or elif statements aren't true.

````python
# 1.4.2
if True:
  print("Hi!")

# 1.4.3
if 2 + 2 == 4:
  print("2 + 2 IS equal to 4.")

# 1.4.4
two = 2
four = 4
if two > four:
  print("2 is bigger than 4.")
elif 2 >= 4:
  print("2 is bigger than or equal to 4.")
else:
  print("2 is not bigger than or equal to 4.")
````

### 1.5 Taking inputs

If you want the code to ask you something and have you enter it, you can use the input statement. Remember, all input you type in would be considered as strings. If you do

```
number = input()
```

even if you type in "5", it would be the string "5", not the integer 5.

If you want to convert a word to a number, you would use the `int()` function. If you want to convert a number to a word, you do the `str()` function like this:
```
# converting str to int
int_version = int("5")

# converting int to str
str_version = str(5)
```

**1.5.2 Revisiting floats**
<br>
In python, numbers without decimals (integers) are different from decimal numbers called "floats". You can pretty much treat them as the same in operators, so this isn't a big issue right now. To convert something to a float, you would do `float()`.

Notice that you **Need** to say `number1 = int(number1)` for number1 to be a number. If you just say `int(number1)`, `int(number1)` is a number, but number1 never changes into a number.

````python
# THE NUMBER MULTIPLIER:
number1 = input()
number2 = input()
# both of these are still words, not numbers, so we need to convert them to numbers

number1 = int(number1)
number2 = int(number2)
# now, they are integers
product = number1 * number2


# you can also make it print it out in words, by turning "product" into a word and adding the other words like below:
product = str(product)
product = "the product is: " + product
print(product)
````

### 1.6 String interpolation


Using the same example as above, another way to do this is to add a "f" in front of a string, then, you can directly add variables inside the string by putting them inside curly brackets (`{}`).

````python
print(f"the product is: {number1 * number2}")
````

**1.6.2 Using `.format`**
<br>
Besides putting the "f" in front of a string, another way to accompolish the same results is to use the `.format()` function. By adding this after a string, we can accompolish the same results. Here's a demonstration:

````python
print("This section teaches you about {}, and today's date is {}.".format("string interpolation", "November 23, 2023"))
````

**1.6.3 Using string interpolation with variables**
<br>
We can also add variable names into the curly brackets. This is useful when we have multiple variables that we want to substitute in:

````python
print("This section teaches you about {topic}, and today's date is {date}.".format(topic = "string interpolation", date = "November 23, 2023"))
````

### 1.7 Comments
