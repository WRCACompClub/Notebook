### 1.5 If statements


An "if" statement runs code if something is true.

Notice how below, it says `if 2 + 2 == 4`. The "==" is different from "="because if you say `age = 5`, it means age **IS** equal to 5. Meanwhile, if you say `age == 5`, this would be asking python if "age is equal to 5?", so it would be a true or false statement.
<br>
<br>
Besides if, there is also "*`elif`*", which just mean "else if". That means if the above if statement isn't true, the code will check this "*`elif`*" statement. You can link as many "*`elif`*" statements to an if statement as you want.
<br>

Besides if and elif, there is also "*`else`*", which only runs if all of the above if or elif statements aren't true.

````python
# 1.5.2
if True:
  print("Hi!")

# 1.5.3
if 2 + 2 == 4:
  print("2 + 2 IS equal to 4.")

# 1.5.4
two = 2
four = 4
if two > four:
  print("2 is bigger than 4.")
elif 2 >= 4:
  print("2 is bigger than or equal to 4.")
else:
  print("2 is not bigger than or equal to 4.")
````

### 1.9 Index

When we want to get individual characters out of a string in python, we can use square brackets to do so. 

For example, let's assign the word "Python" to variable s like this: `s = 'Python'`. Now, if we want to print the first letter "P", we would do this: `print(s[0])`. The same goes if we want to print the last letter "n", we would do this: `print(s[-1])`
<br>

Index applies to strings, lists, dictionaries, and tuples. Python doesn't count the index as how we normally count. Say we are counting from 1 - 5; we start on 1. Python starts at 0, meaning that if we have a list like this:
`my_list = [1, 2, 3, 4, 5]`

And we want to retrieve the integer *4* from the list, we would use `my_list[3]` instead of `my_list[4]`.

### 2.0 For loops

For loop allows the program to run things over and over very quickly. To run something `x` times, we can use this code:

````python
for i in range(10):
    print('Hello!')
````

In the example above, the first line `for i in range(10)` iterates through the `range(10)`, meaning that the code inside the for loop (`print('Hello!')`) will be executed 10 times.
<br>

Here's the structure of a for loop:



The `for` must be lowercase, the first line must end with a colon and the next must start indented (to show that the code is inside the for loop, which tells the program what to execute # of times).

---

We can also use for loop like this:

````python
for i in range(3):
    if i == 0:
        num = int(input('Enter a number: '))
    else:
        num += 1
    print(f'The square of your number is: {num * num}')
````

The example above requires skills from:
*   **1.2**
*   **1.5.4**

Let's break the code apart:

````python
for i in range(3): # repeat the code inside this for loop 3 times
    if i == 0: # ONLY ask the user for input if the loop is running the first time
        num = int(input('Enter a number: '))
    else: # if the loop is in its 2nd or 3rd time, add 1 to the num
        num += 1
    print(f'The square of your number is: {num * num}') # square the num
````

**2.0.2 For loops and indexes**<br>
When we declare a for loop that repeats a specific action for a *x* amount of times, we do this: `for i in range(x):`. The variable `i` in this case serves as an index to the loop. Look closely at this code:

````python
for i in range(10):
    print(i)
````

You will notice that this proves what we've said earlier in *chapter 2.1*. The first index starts at 0, instead of 1. By utilizing indexes and if loops, we can have the program output "first half" and "second half" of the for loop like this:

````python
for i in range(10):
    if i <= 4:
        print('first half')
    else:
        print('second half')
````

As you can see, the code prints exactly *5* "first half" and *5* "second half".
